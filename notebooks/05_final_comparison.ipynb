# 05 Final Comparison

Purpose: combine genAI, specialist, baseline, cost, and runtime results into the final recommendation tables.

```python
import json, pathlib, pandas as pd
result_dir = pathlib.Path('data/results')
rows = []
for path in result_dir.glob('eval_*.json'):
    data = json.loads(path.read_text())
    rows.append({
        'file': path.name,
        'emotion_macro_f1': data['emotions']['macro_f1'],
        'commitment_macro_f1': data['commitment']['macro_f1'],
        'recommendation_macro_f1': data['recommendation']['macro_f1'],
        'commitment_mcc': data['commitment']['mcc'],
        'recommendation_mcc': data['recommendation']['mcc'],
    })
pd.DataFrame(rows).sort_values('emotion_macro_f1', ascending=False)
```
